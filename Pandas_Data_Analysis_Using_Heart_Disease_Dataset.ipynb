QUESTION 1: Dataset Loading & Exploration

```python
import pandas as pd

df = pd.read_csv('/heart.csv')

print("\n1. First 5 rows:")
display(df.head(5))


print("\n2. Last 5 rows:")
display(df.tail(5))

print(f"\n3. Dataset Shape: {df.shape[0]} rows and {df.shape[1]} columns")
```

QUESTION 2: Structure & Missing Values

```python
print("\n1. Column Names:")
print(df.columns.tolist())


print("\n2. Dataset Information:")
df.info()


print("\n3. Missing Values per Column:")
print(df.isnull().sum())
```

QUESTION 3: Categorical Data Analysis

```python

heart_disease_counts = df['target'].value_counts()
print("\n1. Heart Disease Status Counts:")
print(f"Without Heart Disease (0): {heart_disease_counts.get(0, 0)}")
print(f"With Heart Disease (1):    {heart_disease_counts.get(1, 0)}")


gender_counts = df['sex'].value_counts()
print("\n2. Gender Counts:")
print(f"Male (1):   {gender_counts.get(1, 0)}")
print(f"Female (0): {gender_counts.get(0, 0)}")
```

QUESTION 4: Data Selection Using iloc

```python
first_10_rows = df.iloc[0:10]
print("\n1. First 10 rows (iloc[0:10]):")
display(first_10_rows)


age_sex_chol = df.iloc[:, [0, 1, 4]]
print("\n2. Age, Sex, and Cholesterol columns (iloc[:, [0, 1, 4]]):")
display(age_sex_chol.head())


subset_matrix = df.iloc[20:31, 0:5]
print("\n3. Rows 20 to 30, Columns 0 to 4 (iloc[20:31, 0:5]):")
display(subset_matrix)
```

QUESTION 5: Data Filtering Using Conditions

```python

older_than_50 = df[df['age'] > 50]
print(f"\n1. Patients older than 50: {len(older_than_50)} patients found.")
display(older_than_50.head())


hd_and_high_chol = df[(df['target'] == 1) & (df['chol'] > 240)]
print(f"\n2. Heart disease AND cholesterol > 240: {len(hd_and_high_chol)} patients found.")
display(hd_and_high_chol.head())


combined_conditions = df[(df['age'] > 50) & (df['target'] == 1) & (df['chol'] > 240)]
print(f"\n3. Number of patients satisfying all conditions (Age > 50, Target == 1, Chol > 240): {len(combined_conditions)}")
```
