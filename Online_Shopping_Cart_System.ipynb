```python
# Part A: Class Creation

class Shopping_Cart:

    def __init__(self, customer_name, items=None):
        self.customer_name = customer_name
        self.items = list(items) if items is not None else []

    # Part B: Core Methods

    def add_item(self, price):
        if price <= 0:
            print(f"Error: Price must be positive. '{price}' rejected.")
            return
        self.items.append(price)
        print(f"Added item {price:.2f} to {self.customer_name} cart.")

    def remove_item(self, price):
        if price in self.items:
            self.items.remove(price)
            print(f"Removed item {price:.2f} from {self.customer_name} cart.")
        else:
            print(f"Item {price:.2f} not found in {self.customer_name} cart.")

    def calculate_total(self):
        return sum(price for price in self.items)

    def apply_discount(self):
        total = self.calculate_total()
        if total >= 3000:
            return total * 0.90
        return total

    # Part C: Display & Utility Methods

    def display_cart(self):
        total = self.calculate_total()
        final = self.apply_discount()

        print(f"Customer : {self.customer_name}")
        print("Items :")
        if self.items:
            for i, price in enumerate(self.items, start = 1):
                print(f" {i}. {price:.2f}")
        else:
            print("   (cart is empty)")
        print(f"Total : {total:.2f}")
        print(f"After Discount : {final:.2f}")
        print(f"Item Count : {len(self.items)}")


    def clear_cart(self):
        self.items = []
        print(f"{self.customer_name} cart has been cleared.")

    # Combine carts with '+'

    def __add__(self, other):
        combined_name = f"{self.customer_name} & {other.customer_name}"
        combined_items = self.items + other.items
        return Shopping_Cart(combined_name, combined_items)


# Part D: Object Usage & Output

if __name__ == "__main__":
    cart1 = Shopping_Cart("Rahat")
    cart2 = Shopping_Cart("Eva")


    cart1.add_item(1500)
    cart1.add_item(2000)
    cart1.add_item(500)

    cart2.add_item(1000)
    cart2.add_item(2500)


    cart1.remove_item(500)
    cart2.remove_item(9999)


    cart1.display_cart()
    cart2.display_cart()


    combined_cart = cart1 + cart2
    combined_cart.display_cart()
```
