# Deep Learning Project: Handwritten Digit Classifier (MNIST)

## Objective
Build and evaluate a deep learning model that classifies handwritten digits (0–9) using the MNIST dataset.

This notebook covers:
- Dataset loading
- Data preprocessing
- Deep neural network architecture
- Batch Normalization
- Dropout regularization
- Model training with validation split
- Accuracy/loss visualization
- Final test evaluation


```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, BatchNormalization, Dropout
from tensorflow.keras.optimizers import Adam

print('TensorFlow version:', tf.__version__)
```

## 1. Dataset Loading
Load the MNIST dataset directly from TensorFlow/Keras.

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()

print('Training images shape:', x_train.shape)
print('Training labels shape:', y_train.shape)
print('Test images shape:', x_test.shape)
print('Test labels shape:', y_test.shape)
```

## 2. Visualize Sample Images
Display a few handwritten digit samples from the training dataset.

```python
plt.figure(figsize=(10, 4))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i], cmap='gray')
    plt.title(f'Label: {y_train[i]}')
    plt.axis('off')
plt.tight_layout()
plt.show()
```

## 3. Preprocessing
- Normalize pixel values from 0–255 to 0–1.
- Flatten each 28×28 image into a 784-dimensional vector because a dense neural network is used.

```python
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

x_train = x_train.reshape(x_train.shape[0], 28 * 28)
x_test = x_test.reshape(x_test.shape[0], 28 * 28)

print('Processed training shape:', x_train.shape)
print('Processed test shape:', x_test.shape)
print('Minimum pixel value:', x_train.min())
print('Maximum pixel value:', x_train.max())
```

## 4. Model Architecture
The model contains three hidden dense layers, Batch Normalization, and Dropout regularization.

```python
model = Sequential([
    tf.keras.Input(shape=(784,)),

    Dense(256, activation='relu'),
    BatchNormalization(),
    Dropout(0.30),

    Dense(128, activation='relu'),
    BatchNormalization(),
    Dropout(0.30),

    Dense(64, activation='relu'),
    Dropout(0.20),

    Dense(10, activation='softmax')
])

model.summary()
```

## 5. Compile the Model
Sparse categorical crossentropy is used because the target labels are integer class values from 0 to 9.

```python
model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

## 6. Training
Train for multiple epochs using a validation split.

```python
history = model.fit(
    x_train,
    y_train,
    epochs=15,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)
```

## 7. Visualization: Training vs Validation Accuracy

```python
plt.figure(figsize=(8, 5))
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.title('Training vs Validation Accuracy')
plt.legend()
plt.grid(True)
plt.show()
```

## 8. Visualization: Training vs Validation Loss

```python
plt.figure(figsize=(8, 5))
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Training vs Validation Loss')
plt.legend()
plt.grid(True)
plt.show()
```

## 9. Evaluation
Evaluate the trained model on the unseen MNIST test set.

```python
test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)

print(f'Test Loss: {test_loss:.4f}')
print(f'Test Accuracy: {test_accuracy * 100:.2f}%')
```

## 10. Prediction Examples
Use the trained model to predict a few test images.

```python
predictions = model.predict(x_test[:10], verbose=0)
predicted_labels = np.argmax(predictions, axis=1)

plt.figure(figsize=(12, 5))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_test[i].reshape(28, 28), cmap='gray')
    plt.title(f'True: {y_test[i]}\nPred: {predicted_labels[i]}')
    plt.axis('off')
plt.tight_layout()
plt.show()
```

## Conclusion
A fully connected deep neural network was trained on the MNIST handwritten digit dataset. The input data was normalized and flattened, and Batch Normalization and Dropout were used to improve training stability and reduce overfitting. Training/validation accuracy and loss were visualized, and the final model was evaluated on the test dataset.
